### Sección seleccionada del artículo

Del artículo *CO₂ emissions* de Our World in Data (Ritchie & Roser, 2020) el equipo eligió la sección **"Per capita CO₂ emissions"**. Allí se afirma que:

- las emisiones per cápita se calculan dividiendo las emisiones totales de un país entre su población, y se usan cifras **basadas en la producción** (no en el consumo);
- existen **desigualdades muy grandes** entre países: los mayores emisores per cápita son productores de petróleo de Oriente Medio (Qatar, Emiratos, Baréin, Kuwait), mientras que los países más pobres del África subsahariana emiten alrededor de **0,1 toneladas por persona al año**;
- hay una **relación fuerte entre ingreso y emisiones per cápita**, aunque países con niveles de vida parecidos pueden emitir muy distinto según su matriz energética (por ejemplo, Francia frente a Alemania o Estados Unidos).

La pregunta que guía el proyecto es: **¿qué tanto explica el ingreso (PIB per cápita) las emisiones de CO₂ per cápita de los países?**

## 2. Diccionario de las variables que vamos a usar

El conjunto completo tiene 79 columnas. Para la sección de emisiones per cápita se seleccionan solo las necesarias. Las unidades se toman del *codebook* oficial:

## 3. Exploración inicial: tipos de datos y valores faltantes

````python
df = df_crudo[COLUMNAS].copy()

resumen = pd.DataFrame({
    "tipo": df.dtypes.astype(str),
    "faltantes": df.isna().sum(),
    "% faltantes": (df.isna().mean() * 100).round(1),
})
display(resumen)
````

Hay muchos faltantes porque el archivo arranca en 1750, cuando casi ningún país tenía registros. Además, la columna `country` mezcla **países** con **agregados** (continentes, grupos de ingreso, la Unión Europea, el mundo). Si no se separan, se contaría dos veces la misma emisión.

````python
agregados = df[df["iso_code"].isna()]["country"].unique()
print(f"Entidades sin código ISO: {len(agregados)}")
print(sorted(agregados))
````

Casi todas son agregados. La excepción es **Kosovo**, que sí es un país pero no tiene código ISO de tres letras en el archivo. Se le asigna el código `XKX`, que usa el Banco Mundial, para no perderlo.

## 4. Separación de países y agregados

````python
df.loc[df["country"] == "Kosovo", "iso_code"] = "XKX"

# Agregados útiles que se guardan aparte para el análisis exploratorio
mundo = df[df["country"] == "World"].copy()
GRUPOS = ["High-income countries", "Upper-middle-income countries",
          "Lower-middle-income countries", "Low-income countries"]
grupos = df[df["country"].isin(GRUPOS)].copy()

# Países: filas con código ISO de tres letras
paises = df[df["iso_code"].notna() & (df["iso_code"].str.len() == 3)].copy()

print("Filas del mundo:", len(mundo))
print("Filas de grupos de ingreso:", len(grupos))
print("Filas de países:", len(paises), "| países distintos:", paises["country"].nunique())
````

## 5. Filtro temporal

La sección del artículo compara las emisiones per cápita **actuales** entre países. Para el análisis de países se conserva el periodo **1950–2025**: antes de 1950 muy pocos países tienen datos y los que existen son estimaciones gruesas.

La serie del mundo se conserva completa (desde 1750), porque el artículo también muestra la evolución histórica desde la Revolución Industrial.

````python
filas_antes = len(paises)
paises = paises[paises["year"] >= 1950].copy()
print(f"Filas antes: {filas_antes} | después del filtro: {len(paises)}")
print("Faltantes en co2_per_capita (1950-2025):", paises["co2_per_capita"].isna().sum())
````

## 6. Duplicados

````python
duplicados = paises.duplicated(subset=["country", "year"]).sum()
print("Filas duplicadas (mismo país y año):", duplicados)
paises = paises.drop_duplicates(subset=["country", "year"])
````

## 7. Tratamiento de valores faltantes

Decisiones, variable por variable:

| Variable | Decisión | Por qué |
|---|---|---|
| `co2_per_capita` | Se **eliminan** las filas sin dato | Es la variable de interés: imputarla inventaría justo lo que se quiere analizar. |
| `population` | Se eliminan las filas sin dato | Sin población no hay per cápita. |
| `gdp` | Se **conserva el faltante** (NaN) | Solo hace falta para la regresión. El notebook 3 usa únicamente los países que lo tienen. |
| `coal_co2`, `oil_co2`, `gas_co2`, `cement_co2`, `flaring_co2` | Se conserva el faltante | Muchos países no producen cemento ni queman gas: no se reemplaza por cero porque no hay forma de distinguir un cero real de un dato que no se midió. |

````python
antes = len(paises)
paises = paises.dropna(subset=["co2_per_capita", "population"])
print(f"Filas eliminadas por falta de co2_per_capita o población: {antes - len(paises)}")
print(f"Filas que quedan: {len(paises)}")

display((paises.isna().mean() * 100).round(1).rename("% faltantes").to_frame())
````

## 8. Verificación de consistencia

Se revisan tres cosas:

1. Que no haya emisiones ni población **negativas**.
2. Que `co2_per_capita` coincida con `co2 / population`. Como `co2` está en millones de toneladas, la cuenta es `co2 × 1 000 000 / population`.
3. Que la suma de emisiones de todos los países sea parecida al total mundial (las diferencias son normales, porque el total mundial incluye la aviación y el transporte marítimo internacionales, que no se asignan a ningún país).

````python
for col in ["co2", "co2_per_capita", "population"]:
    print(f"Valores negativos en {col}:", (paises[col] < 0).sum())

calculado = paises["co2"] * 1e6 / paises["population"]
dif_relativa = ((calculado - paises["co2_per_capita"]).abs() / paises["co2_per_capita"]) * 100
print(f"\nDiferencia relativa entre co2_per_capita y co2/población:")
print(f"  mediana = {dif_relativa.median():.3f} %  |  máxima = {dif_relativa.max():.2f} %")
print("  filas con diferencia mayor al 1 %:", (dif_relativa > 1).sum())

suma = paises.groupby("year")["co2"].sum()
total = mundo.set_index("year")["co2"]
comparacion = pd.DataFrame({"suma_paises": suma, "total_mundo": total}).dropna()
comparacion["% cubierto"] = (comparacion["suma_paises"] / comparacion["total_mundo"] * 100).round(1)
display(comparacion.loc[[1950, 1980, 2000, 2010, 2020, 2024, 2025]])
````

No hay negativos, y los países suman entre el 96 % y el 98 % de las emisiones mundiales: el resto es la aviación y el transporte marítimo internacionales.

La columna per cápita coincide con `co2 / población`: la diferencia típica (mediana) es de apenas 0,02 %. Las diferencias más grandes aparecen cuando las emisiones son diminutas. Por ejemplo, Bután en 1970 tiene `co2 = 0.004` millones de toneladas, y como esa columna solo trae tres decimales, el redondeo pesa mucho en el cociente. No son errores del dato per cápita, sino del redondeo de la columna `co2`; por eso se trabaja con `co2_per_capita` tal como la publica Our World in Data.

## 8.1 Microterritorios y ceros sospechosos

Revisando los valores más altos aparece un problema de calidad: **Sint Maarten registra 1 062 toneladas de CO₂ por persona en 1956**, unas 75 veces lo que emite hoy un estadounidense. Con solo unos 2 000 habitantes, una sola refinería basta para disparar el indicador. En territorios tan pequeños el per cápita refleja instalaciones puntuales, no el comportamiento de una población, y distorsionaría tanto las gráficas como la regresión.

**Decisión:** se excluyen del análisis los territorios cuya población **nunca supera los 100 000 habitantes**. Los datos siguen disponibles en el archivo original.

También hay **14 filas con emisión per cápita exactamente igual a 0** en países con cientos de miles o millones de habitantes (Kosovo 1994–1999, Timor Oriental 1994–1997, Kuwait 1952–1953, Arabia Saudita 1951–1952). Un país así no emite exactamente cero: son registros vacíos que vinieron como 0. Se tratan como faltantes y se eliminan.

````python
pob_max = paises.groupby("country")["population"].max()
micro = pob_max[pob_max < 100_000].index
print(f"Microterritorios excluidos ({len(micro)}):")
print(", ".join(sorted(micro)))

antes = len(paises)
paises = paises[~paises["country"].isin(micro)]
ceros = (paises["co2_per_capita"] == 0).sum()
paises = paises[paises["co2_per_capita"] > 0]
print(f"\nFilas eliminadas por microterritorios: {antes - len(paises) - ceros}")
print(f"Filas eliminadas por ceros sospechosos: {ceros}")
print(f"Filas que quedan: {len(paises)} | países: {paises['country'].nunique()}")
print(f"Nuevo máximo de co2_per_capita: {paises['co2_per_capita'].max():.1f} t "
      f"({paises.loc[paises['co2_per_capita'].idxmax(), 'country']}, "
      f"{paises.loc[paises['co2_per_capita'].idxmax(), 'year']})")
````

## 9. Variables derivadas

- `gdp_per_capita`: PIB / población, en dólares internacionales de 2011 por persona.
- `log_gdp_pc` y `log_co2_pc`: logaritmos en base 10. Ambas variables tienen distribuciones muy sesgadas (pocos países muy ricos o muy emisores), y el logaritmo las vuelve más simétricas. Esto se usa en la regresión del notebook 3.
- `fosil_co2`: suma de carbón, petróleo y gas.

````python
paises["gdp_per_capita"] = paises["gdp"] / paises["population"]
paises["log_gdp_pc"] = np.log10(paises["gdp_per_capita"])
paises["log_co2_pc"] = np.log10(paises["co2_per_capita"].where(paises["co2_per_capita"] > 0))
paises["fosil_co2"] = paises[["coal_co2", "oil_co2", "gas_co2"]].sum(axis=1, min_count=1)

print("Filas con co2_per_capita igual a 0 (no tienen logaritmo):", (paises["co2_per_capita"] == 0).sum())
display(paises[["co2_per_capita", "gdp_per_capita", "energy_per_capita"]].describe().round(2))
````

## 10. Valores atípicos

Se usa el criterio del **rango intercuartílico (IQR)** sobre las emisiones per cápita del año más reciente con datos completos (2024). Un valor es atípico si está por encima de Q3 + 1,5·IQR.

Los atípicos **no se eliminan**: son países reales, y el propio artículo los menciona como los mayores emisores per cápita. Solo se marcan con una columna, para tenerlos en cuenta en la visualización y en el modelo.

````python
anio = 2024
corte = paises[paises["year"] == anio]["co2_per_capita"]
q1, q3 = corte.quantile([0.25, 0.75])
limite = q3 + 1.5 * (q3 - q1)
print(f"Q1 = {q1:.2f} t | Q3 = {q3:.2f} t | límite superior = {limite:.2f} t por persona")

paises["atipico_co2_pc"] = paises["co2_per_capita"] > limite
atipicos = (paises[(paises["year"] == anio) & paises["atipico_co2_pc"]]
            [["country", "co2_per_capita", "population"]]
            .sort_values("co2_per_capita", ascending=False))
print(f"\nPaíses atípicos en {anio}: {len(atipicos)}")
display(atipicos.round(2).reset_index(drop=True))
````

Los atípicos coinciden con lo que describe el artículo:

- **Productores de petróleo y gas** (Qatar, Baréin, Brunéi, Kuwait, Trinidad y Tobago, Emiratos, Arabia Saudita, Omán). La mayoría tiene poca población, así que, aunque emiten mucho por persona, su peso en el total mundial es bajo.
- **Estados Unidos, Australia y Canadá**, que según el artículo emiten alrededor de tres veces el promedio mundial. Aparecen junto a Kazajistán y Rusia, economías muy dependientes del carbón y del gas.

En la serie histórica también aparece un valor extremo real: **Kuwait en 1991 (364,9 t por persona)**. No es un error: corresponde a los incendios de los pozos petroleros durante la Guerra del Golfo. Se conserva, pero el análisis se concentra en años recientes.

8. Se crearon el PIB per cápita, los logaritmos y las emisiones fósiles.
